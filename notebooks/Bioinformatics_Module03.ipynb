```python
#!/usr/bin/python3

sequences = {
    "H.neandertalensis":"ACGTCGCTCTGCTA",
    "H.sapiens":"ACGTCACTCTGCTA"
    }

print(sequences)
```

```python
fasta_filename = "hominid_sequences.fasta"

with open(fasta_filename, "w") as f:
  for header, sequence in sequences.items():
    f.write(f">{header}\n")
    f.write(f"{sequence}\n")

print(f"FASTA file '{fasta_filename}' created successfully.")

with open(fasta_filename, "r") as f:
  print("\nContent of the FASTA file:")
  print(f.read())
```

```python
all_words = []
with open("practice.txt", "r") as data:
  for line in data:
    all_words.extend(line.split())

word_counts = {}

unique_words = sorted(list(set(all_words)))

for word in unique_words:
  word_counts[word] = all_words.count(word)

for word, count in word_counts.items():
  print(f"{word} {count}")
```

```python
from Bio.SeqIO import parse

for seq_record in parse("practice_9.txt", "fasta"):
  print(f"Sequence ID: {seq_record.id}")
  print(f"Sequence: {seq_record.seq}")
  print(f"Length: {len(seq_record.seq)}")

  g_count = seq_record.seq.count('G')
  c_count = seq_record.seq.count('C')
  total_bases = len(seq_record.seq)

  if total_bases > 0:
    gc_content = ((g_count + c_count) / total_bases) * 100
    print(f"GC Content: {gc_content:.2f}%")
  else:
    print("GC Content: N/A (empty sequence)")
  print("-----------------------------------")
```
